```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
train_loss_basic = np.load("./data/gcn_train_loss_basic.npy")
dnn_train_loss_hpt = np.load("./data/dnn_train_loss_hpt.npy")
gcn_train_loss_hpt = np.load("./data/gcn_train_loss_hpt.npy")
gat_train_loss_hpt = np.load("./data/gat_train_loss_hpt.npy")
graphsage_train_loss_hpt = np.load("./data/graphsage_train_loss_hpt.npy")
```

```python
# 1. 예시 데이터 생성
x = np.arange(0,train_loss_basic.shape[0])

# 2. Figure 생성 및 그래프 그리기
plt.figure(figsize=(8, 5))

# 각 plot에 label을 지정하는 것이 핵심입니다.
plt.plot(x, train_loss_basic, label='Basic', color='blue', linewidth=2)
plt.plot(x, gcn_train_loss_hpt, label='HPT', color='red', linestyle='--', linewidth=2)

# # 3. 타이틀 및 x, y축 레이블 추가
# plt.title('Comparison of Two Lines', fontsize=14, pad=15)
plt.xlabel('Epoch', fontsize=12)
plt.ylabel('Loss', fontsize=12)

# 4. 범례(Legend) 표시 및 그리드 추가
# loc='upper right' 외에도 'upper left', 'lower right', 'best' 등으로 위치 조절이 가능합니다.
plt.legend(loc='upper right', fontsize=15)
plt.grid(True, linestyle=':', alpha=0.6)

# 5. 그래프 출력
plt.tight_layout()
plt.show()
```
